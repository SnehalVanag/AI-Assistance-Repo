```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

datagen = ImageDataGenerator(
    rescale=1./255
)

print("Image preprocessing ready!")
```

```python
train_data = datagen.flow_from_directory(
    dataset_path,
    target_size=(150, 150),
    batch_size=4,
    class_mode='binary'
)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense, Dropout

model = Sequential([

    # First Convolution Layer
    Conv2D(32, (3, 3), activation='relu',
           input_shape=(150, 150, 3)),
    MaxPooling2D(2, 2),

    # Second Convolution Layer
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),

    # Third Convolution Layer
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),

    # Convert feature maps into one-dimensional data
    Flatten(),

    # Fully Connected Layer
    Dense(128, activation='relu'),

    # Reduce overfitting
    Dropout(0.5),

    # Output Layer
    Dense(1, activation='sigmoid')
])

print("CNN Model Created Successfully!")
```

```python
import os

for folder in ['cat', 'dog', 'other']:
    path = os.path.join(dataset_path, folder)
    print(folder, "images:", len(os.listdir(path)))
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

datagen = ImageDataGenerator(
    rescale=1./255
)

train_data = datagen.flow_from_directory(
    dataset_path,
    target_size=(150, 150),
    batch_size=4,
    class_mode='categorical'
)

print("3-Class Dataset Ready!")
print("Class Indices:", train_data.class_indices)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense, Dropout

model = Sequential([
    Input(shape=(150, 150, 3)),

    # Convolution Layer 1
    Conv2D(32, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),

    # Convolution Layer 2
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),

    # Convolution Layer 3
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),

    # Flatten
    Flatten(),

    # Fully Connected Layer
    Dense(128, activation='relu'),
    Dropout(0.5),

    # 3 Classes
    Dense(3, activation='softmax')
])

print("3-Class CNN Model Created Successfully!")
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("Model compiled successfully!")
```

```python
history = model.fit(
    train_data,
    epochs=10
)
```

```python
from tensorflow.keras.utils import load_img, img_to_array
import numpy as np
import matplotlib.pyplot as plt

test_image_path = '/content/drive/MyDrive/bs.jpeg'

image = load_img(
    test_image_path,
    target_size=(150, 150)
)

plt.imshow(image)
plt.axis('off')
plt.show()

image_array = img_to_array(image)
image_array = image_array / 255.0
image_array = np.expand_dims(image_array, axis=0)

prediction = model.predict(image_array)

print("Prediction values:", prediction[0])
```

```python
prediction = model.predict(image_array)

cat_probability = prediction[0][0]
dog_probability = prediction[0][1]
other_probability = prediction[0][2]

print("Cat Probability:", round(cat_probability * 100, 2), "%")
print("Dog Probability:", round(dog_probability * 100, 2), "%")
print("Other Probability:", round(other_probability * 100, 2), "%")

if cat_probability > dog_probability and cat_probability > other_probability:
    print("Prediction: CAT 🐱")

elif dog_probability > cat_probability and dog_probability > other_probability:
    print("Prediction: DOG 🐶")

else:
    print("Prediction: OTHER ❓")
```
